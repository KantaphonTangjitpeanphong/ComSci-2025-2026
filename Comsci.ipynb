Task 1 = Scores for events (15Marks)

```python
CompetitorName = ["August","chris","Phoom"]
CompetitorScore = [[1,],
                   [2,],
                   [3,],]
point = []
```

```python
print("score input for first event")
for i in range(3):
    print(i)
    print(f"input the score of {CompetitorName[i]}")
    score = int(input())
    while True:
        try: 
            sc = int(score)
            if sc < 0 or sc > 100:
                print("score must be positive and below 100")
                score = int(input())
            else:
                break
        except:
            print("score must be a number")
        



    CompetitorScore[i].append(sc)
    print(CompetitorScore)
```

```python
print("score input for second event")

for n in range(3):
    print(n)
    print(f"input the score of {CompetitorName[n]}")
    score = int(input())
    while True:
        try: 
            sc = int(score)
            if sc < 0 or sc > 100:
                print("score must be positive and below 100")
                score = int(input())
            else:
                break
        except:
            print("score must be a number")
    CompetitorScore[n].append(sc)
    print(CompetitorScore)
```

```python
for g in range(2):
    point.append(CompetitorScore[g][1] + CompetitorScore[g][2])
    print(point)
```

```python
max = 0
for h in range(3):
    if CompetitorScore[1][h] > max:
        max = CompetitorScore[1][h]
        index = h
print(f"the winner of event1 is {CompetitorName[index]} with score {max}")      
```

```python
print("find max of event2")

max = 0
for j in range(3):
    if CompetitorScore[2][j] > max:
        max = CompetitorScore[2][j]
        index = j
print(f"the winner of event2 is {CompetitorName[index]} with score {max}")
```

```python
print("finding champion")
for k in range(2):
    if point[k] > max:
        max = point[k]
        index = k
print(f"the champion is {CompetitorName[index]} with score {max}")
print(f"medal awarded to {CompetitorName[index]}")
```
